### Stress Level (1–10)
A simple self‑rating tool where a person estimates how stressed they feel right now. It’s subjective, but using clear anchors helps keep ratings consistent.

1 – Completely relaxed  
No noticeable stress. Calm, clear thinking, body relaxed.

2 – Very mild tension  
Slight pressure but easily manageable.

3 – Low stress  
Aware of some demands, but still comfortable and functioning well.

4 – Noticeable stress  
You feel somewhat pressured or distracted.

5 – Moderate stress  
Stress is clearly present; concentration takes more effort.

6 – Elevated stress  
Feeling tense or irritable; performance may start to drop.

7 – High stress  
Difficult to focus, physical symptoms may appear (tight shoulders, faster heart rate).

8 – Very high stress  
Overwhelmed; thinking becomes less clear and emotions stronger.

9 – Extreme stress  
Close to panic or shutdown; functioning is very difficult.

10 – Maximum stress / crisis  
Panic, breakdown, or inability to cope.

How people usually measure it:
- Self‑check: “Where am I right now on the 1–10 scale?”
- Body cues: heart rate, muscle tension, breathing, headaches.
- Mental cues: racing thoughts, inability to focus, irritability.
- Behavior cues: avoidance, snapping at people, mistakes.

Many therapists and coaches ask clients to rate stress before and after an activity (meeting, workout, breathing exercise) to see what helps.

### Sweating Level (1–5)

1 – None / Dry  
Skin is dry. No noticeable sweating.

2 – Slight  
Very light moisture on skin (forehead, palms, or underarms), but not visible to others.

3 – Moderate  
Clearly noticeable sweat. Skin feels damp; small sweat spots may appear on clothing.

4 – Heavy  
Visible sweating and larger sweat patches on clothing. Sweat may drip or require wiping.

5 – Extreme / Profuse  
Constant dripping or soaking clothing. Sweat runs down face or body.

This type of scale is often used in exercise studies, heat stress monitoring, or medical observations to quickly rate sweating intensity.

### Diet Quality (1–10)

1 – Very poor  
Mostly ultra‑processed foods, sugary drinks, fast food, little to no fruits or vegetables.

2 – Poor  
Frequent junk food and refined carbs; very low nutrient variety.

3 – Below average  
Some real food, but still heavy in processed items and added sugars.

4 – Somewhat unhealthy  
Occasional fruits/vegetables but inconsistent balance; many refined foods.

5 – Average  
Mixed diet: some healthy meals, some unhealthy ones.

6 – Fairly good  
Regular fruits/vegetables and home‑cooked meals, but still room for improvement.

7 – Good

8 – Very good  
Mostly whole foods, good nutrient balance, limited processed foods.

9 – Excellent  
Highly balanced diet with plenty of whole foods, healthy fats, lean proteins, fiber.

10 – Optimal  
Consistently nutrient‑dense, varied diet aligned with strong nutrition guidelines (lots of whole foods, minimal processed items).

People often estimate this based on factors like:
- fruit and vegetable intake  
- whole vs processed foods  
- sugar and sugary drink consumption  
- protein quality  
- variety of nutrients

### Difference Between Stress Levels and Anxiety Levels

Stress and anxiety are related but not the same.

Stress
- Usually a response to a specific external pressure (deadline, exam, argument, workload).
- Often temporary and improves when the situation passes.
- More tied to current demands.
- Symptoms: tension, irritability, trouble concentrating, headaches, fatigue.
- Example: feeling stressed because you have a project due tomorrow.

Anxiety
- More about worry, fear, or anticipation of possible threats, sometimes without a clear trigger.
- Can persist even when no immediate stressor exists.
- Often involves excessive worrying about the future.
- Symptoms: constant worry, restlessness, racing thoughts, rapid heartbeat, sleep problems.
- Example: worrying for days that you might fail something even when nothing immediate is happening.

Simple way to think about it
- Stress: pressure from a situation.  
- Anxiety: the internal feeling of fear or worry, sometimes continuing after the stressor is gone.

They often overlap: prolonged stress can lead to anxiety, and anxiety can make normal stress feel much stronger.

People often judge their number using:

## Well-known Psychology/Health KPIs Derived from Ratios

## 1. Anxiety-to-Stress Ratio
**Formula:** `Anxiety Level / Stress Level`

Measures how much anxiety a person experiences relative to their perceived stress. A ratio >1 suggests anxiety is disproportionate to stressors (common in GAD).

## 2. Sleep Efficiency Index (proxy)
**Formula:** `Sleep Hours / 8`

Normalized sleep adequacy score. Values <0.75 are clinically associated with anxiety and mood disorders.

## 3. Caffeine-to-Sleep Ratio
**Formula:** `Caffeine Intake / Sleep Hours`

High caffeine relative to sleep is a known anxiety amplifier. Used in behavioral health screening.

## 4. Physical Activity-to-Stress Ratio
**Formula:** `Physical Activity / Stress Level`

Reflects coping capacity. Higher activity relative to stress = better resilience. Used in lifestyle medicine.

## 5. Therapy Engagement Ratio
**Formula:** `Therapy Sessions / Anxiety Level`

How much therapeutic support someone is receiving per unit of anxiety. Low values flag undertreated cases.

## 6. Autonomic Arousal Index
**Formula:** `Heart Rate / Breathing Rate`

A known physiological ratio. Normal resting ratio ≈ 4:1 (72 bpm / 18 breaths). Elevated values indicate sympathetic nervous system activation (fight-or-flight).

## 7. Alcohol-to-Sleep Ratio
**Formula:** `Alcohol Consumption / Sleep Hours`

Alcohol disrupts sleep architecture; this ratio is used in substance-related anxiety assessment.

## 8. Substance Load vs. Diet Quality
**Formula:** `(Caffeine Intake + Alcohol Consumption) / Diet Quality`

A composite lifestyle burden score — higher values indicate poor health behaviors relative to diet quality.

## 9. Stress-to-Coping Ratio
**Formula:** `Stress Level / (Physical Activity + Therapy Sessions + 1)`

+1 avoids division by zero. Higher = worse coping relative to stressors. Directly maps to the Stress-Coping Balance model in clinical psychology.

---

The most psychologically validated of these are **#1 (Anxiety/Stress)**, **#6 (HR/BR)**, and **#9 (Stress-Coping)** — they have direct parallels in clinical assessment literature.

### Normality

```python
def normality_checker(df, columns=None, alpha=0.05) -> pd.DataFrame:
    
    if columns is None:
        columns = df.select_dtypes(include='number').columns
    
    results = []
    
    for col in columns:
        clean_data = df[col].dropna()
        n = len(clean_data)
        
        if n < 3:
            results.append([col, n, None, "Not enough data"])
            continue
        
        stat, p_value = shapiro(clean_data)
        
        result = "Normal" if p_value > alpha else "Not normal"
        
        results.append([col, n, p_value, result])
    
    result_df = pd.DataFrame(
        results,
        columns=["Column", "Sample Size", "p-value", "Result"]
    )
    
    return result_df
```

### NaN percentage

```python
def nan_percentage(df : pd.DataFrame) -> pd.DataFrame:
    total_rows = len(df)
    missing_count = [df[col].isna().sum() for col in df.columns]
    nan_percentage = [(df[col].isna().sum() / total_rows) * 100 for col in df.columns]
    result = pd.DataFrame({'Missing Count': missing_count, 'Nan Percent': nan_percentage}, index=df.columns)
    sorted_result = result.sort_values(by ='Nan Percent', ascending = False)
    return sorted_result
```

```python
def check_invalid_nan_intersection(data):
    conditions = {
        'Sleep Hours': data['Sleep Hours'] < 0,
        'Physical Activity (hrs/week)': data['Physical Activity (hrs/week)'] < 0,
        'Alcohol Consumption (drinks/week)': data['Alcohol Consumption (drinks/week)'] < 0,
        'Stress Level (1-10)': data['Stress Level (1-10)'] > 10,
    }

    invalid_count = pd.DataFrame(conditions).sum(axis=1)
    nan_count = data.isna().sum(axis=1)

    mask = (invalid_count > 1) & (nan_count > 2)

    result = data[mask].copy()
    result['Invalid_Count'] = invalid_count[mask]
    result['NaN_Count'] = nan_count[mask]
    result['Total_Issues'] = result['Invalid_Count'] + result['NaN_Count']

    print('rows with both NaN and invalid values:', len(result))
    return result
```

### Impute NaNs

```python
def fill_numeric(data, cols):
    data = data.copy()
    for c in cols:
        if data[c].isnull().sum() > 0:
            med = data[c].median()
            print(c, '->', data[c].isnull().sum(), 'filled with median', med)
            data[c] = data[c].fillna(med)
    return data

def fill_categorical(data, cols):
    data = data.copy()
    for c in cols:
        if data[c].isnull().sum() > 0:
            m = data[c].mode()[0]
            print(c, '->', data[c].isnull().sum(), 'filled with mode', m)
            data[c] = data[c].fillna(m)
    return data
```

## Outliers, Binning & Reclassifying

### Outliers

```python
def outlier_detector(col, method='iqr', k=1.5, z_thresh=3, clip=False) -> pd.Series:
    if method == 'iqr':
        q1, q3 = col.quantile(0.25), col.quantile(0.75)
        iqr = q3 - q1
        lower = q1 - k * iqr
        upper = q3 + k * iqr
        mask = (col < lower) | (col > upper)
        
        if clip:
            return col.clip(lower, upper)
        return mask

    elif method == 'zscore':
        z = zscore(col).abs()
        mask = z > z_thresh
        
        if clip:
            mean, std = col.mean(), col.std()
            lower = mean - z_thresh * std
            upper = mean + z_thresh * std
            return col.clip(lower, upper)
        return mask

    elif method == 'intersection':
        iqr_mask = outlier_detector(col, 'iqr', k=k, clip=False)
        z_mask = outlier_detector(col, 'zscore', z_thresh=z_thresh, clip=False)
        mask = iqr_mask & z_mask
        
        if clip:
            q1, q3 = col.quantile(0.25), col.quantile(0.75)
            iqr_val = q3 - q1
            iqr_lower = q1 - k * iqr_val
            iqr_upper = q3 + k * iqr_val
            result = col.copy()
            result[mask] = result[mask].clip(iqr_lower, iqr_upper)
            return result
        return mask
```

### Binning

Binning turns a continuous variable into a smaller number of meaningful groups — useful when the raw
scale is noisy but a coarser grouping tells a clearer story (e.g. age brackets instead of exact age).

```python
def binning(data, col, method='custom', bins=5, thresholds=None, labels=None):
    
    df = data.copy()
    
    new_col = col + "_bin"
    
    if method == "equal width":
        df[new_col] = pd.cut(df[col], bins=bins,  labels=labels)
    
    elif method == "quantile":
        df[new_col] = pd.qcut(df[col], q=bins,  labels=labels)
    
    elif method == "custom":
        df[new_col] = pd.cut(df[col], bins=thresholds, labels=labels)
    
    print("column:", new_col)
    print("number of groups:", df[new_col].nunique())
    
    return df
```

### Reclassifying categorical variables

### Confidence Interval Analysis

- In this part, we used confidence intervals for numerical columns.
- The goal was to estimate the possible population mean based on our sample.

- We used confidence intervals for variables such as:
  - `Sleep Hours`
  - `Caffeine Intake (mg/day)`
  - `Physical Activity (hrs/week)`

- A confidence interval gives a range of likely values for the true population mean.
- For example, instead of reporting only the sample mean of anxiety level, we reported a 95% confidence interval.

- This makes the result more informative because it also shows uncertainty.

```python
def mean_ci(data, column, confidence=0.95):
    
    x = data[column]
    n = len(x)
    mean = x.mean()
    std = x.std(ddof=1)
    se = std / np.sqrt(n)

    alpha = 1 - confidence
    t_critical = stats.t.ppf(1 - alpha / 2, df=n - 1)

    margin_error = t_critical * se
    lower = mean - margin_error
    upper = mean + margin_error

    result = pd.DataFrame({
        "Variable": [column],
        "Type": "Mean CI",
        "N": n,
        "Sample Mean": round(mean, 3),
        "Sample Std": round(std, 3),
        "Confidence Level": confidence,
        "CI Lower": round(lower, 3),
        "CI Upper": round(upper, 3),
    })

    return result
```

```python
def anova_test(data, multi_cat_col, target_col):

    groups = []

    for group in data[multi_cat_col].unique():

        sample = data[data[multi_cat_col] == group][target_col]

        groups.append(sample)

    f_stat, p = stats.f_oneway(*groups)

    if p < 0.05:
        result = "Significant"
    else:
        result = "Not Significant"

    output = pd.DataFrame({"multi_category": [multi_cat_col],
                          "target": [target_col],
                          "F_stat": f_stat,
                          "P_value": p,
                          "result": result})

    return output
```

### Numeric vs Categorical Analysis
This analysis examines whether there is a statistically significant association between two categorical variables using the Chi-Square Test of Independence.

Null Hypothesis (H₀):** The two categorical variables are independent. There is no significant association between them.
Alternative Hypothesis (H₁):** The two categorical variables are not independent. There is a significant association between them.
If p-value < 0.05, reject the null hypothesis (H₀).
If p-value ≥ 0.05, fail to reject the null hypothesis (H₀).

- To compare a continuous feature against a binary categorical target, we first assess normality
- using the Shapiro-Wilk test. If both groups are normally distributed, we apply an independent
- samples t-test. If not, we attempt a Box-Cox transformation to normalize the data and re-test.
- If normality still cannot be achieved, we fall back to the non-parametric Mann-Whitney U test.

```python
def numeric_vs_categorical(data, col_1, col_2, alpha=0.05):

    numeric_vs_categorical_plot(data, col_1, col_2)

    data = data.copy()
    data[col_2] = np.where(data[col_2] == 1, 'Yes', data[col_2])
    data[col_2] = np.where(data[col_2] == 0, 'No',  data[col_2])

    categories_unique = data[col_2].dropna().unique()
    X1 = data[data[col_2] == categories_unique[0]][col_1].dropna()
    X2 = data[data[col_2] == categories_unique[1]][col_1].dropna()

    shapiro_p0 = stats.shapiro(X1).pvalue
    shapiro_p1 = stats.shapiro(X2).pvalue
    print(f"Shapiro group 1: p = {shapiro_p0:.4f}")
    print(f"Shapiro group 2: p = {shapiro_p1:.4f}")

    p = None  

    if shapiro_p0 > alpha and shapiro_p1 > alpha:
        t, p = stats.ttest_ind(X1, X2)
        print(f"Normal distribution -> t-test")
        print(f"t = {t:.4f}, p = {p:.4f}")

    else:
        try:
            X1_transformed, _ = stats.boxcox(X1 + 1)
            X2_transformed, _ = stats.boxcox(X2 + 1)
            shapiro_p0_bc = stats.shapiro(X1_transformed).pvalue
            shapiro_p1_bc = stats.shapiro(X2_transformed).pvalue

            if shapiro_p0_bc > alpha and shapiro_p1_bc > alpha:
                t, p = stats.ttest_ind(X1_transformed, X2_transformed)
                print(f"Normal after Box-Cox -> t-test")
                print(f"t = {t:.4f}, p = {p:.4f}")
            else:

                u, p = stats.mannwhitneyu(X1, X2, alternative="two-sided")
                print(f"Not normal after Box-Cox ->> Mann-Whitney")
                print(f"Statistic = {u:.4f}, p = {p:.4f}")

        except Exception:
            u, p = stats.mannwhitneyu(X1, X2, alternative="two-sided")
            print(f"Not normal -->> Mann-Whitney")
            print(f"Statistic = {u:.4f}, p = {p:.4f}")

    if p is not None:
        if p < alpha:
            print(f"Significant difference between '{col_1}' groups")
        else:
            print(f"No significant difference between '{col_1}' groups")

    
```

### compare two categorical
This analysis examines whether there is a statistically significant association between two categorical variables using the Chi-Square Test of Independence.
- Null Hypothesis (H₀):** The two categorical variables are independent. There is no significant association between them.
- Alternative Hypothesis (H₁):** The two categorical variables are not independent. There is a significant association between them.
- If **p-value < 0.05**, reject the null hypothesis (H₀).
- If **p-value ≥ 0.05**, fail to reject the null hypothesis (H₀).

```python
def compare_two_categorical(df, col_1, col_2, alpha=0.05):

    two_categorical_plot(df, col_1, col_2)
    
    Data = data.copy()      
    Data[col_2] = np.where(Data[col_2] ==1 ,'yes',Data[col_2])
    Data[col_2]=  np.where(Data[col_2] ==0,'No', Data[col_2])
        
    contingency_table = pd.crosstab(Data[col_1],Data[col_2])

    chi2_stat, p_value, dof, expected_table = chi2_contingency(contingency_table)

    print(f"chi2_stat: {chi2_stat:.4f}")
    print(f"p-value: {p_value:.4f}")
        
    if p_value < alpha:  
        print("Reject the Null Hypothesis")
        print(f"There IS a significant association between '{col_1}' and '{col_2}'.")
    else:
        print("Fail to Reject the Null Hypothesis")
        print(f"There is NO significant association between '{col_1}' and '{col_2}'.")

    
```

#### 5. Stress_Coping_Ratio

```python
physical_act_scaled = (data['Physical Activity (hrs/week)'] - data['Physical Activity (hrs/week)'].min()) /\
                    (data['Physical Activity (hrs/week)'].max() - data['Physical Activity (hrs/week)'].min())
therapy_ses_scaled = (data['Therapy Sessions (per month)'] - data['Therapy Sessions (per month)'].min()) / \
                    (data['Therapy Sessions (per month)'].max() - data['Therapy Sessions (per month)'].min())
data['Stress_Coping_Ratio'] = data['Stress Level (1-10)'] / (physical_act_scaled + therapy_ses_scaled + 1)
```

```python
numeric_vs_categorical(data, 'Stress_Coping_Ratio', 'Target')
```

**Result:** Represents maladaptive coping behavior: - alcohol - smoking - caffeine
