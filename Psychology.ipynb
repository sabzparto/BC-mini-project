### Stress Level (1–10)
A simple self‑rating tool where a person estimates how stressed they feel right now. It’s subjective, but using clear anchors helps keep ratings consistent.

1 – Completely relaxed  
No noticeable stress. Calm, clear thinking, body relaxed.

2 – Very mild tension  
Slight pressure but easily manageable.

3 – Low stress  
Aware of some demands, but still comfortable and functioning well.

4 – Noticeable stress  
You feel somewhat pressured or distracted.

5 – Moderate stress  
Stress is clearly present; concentration takes more effort.

6 – Elevated stress  
Feeling tense or irritable; performance may start to drop.

7 – High stress  
Difficult to focus, physical symptoms may appear (tight shoulders, faster heart rate).

8 – Very high stress  
Overwhelmed; thinking becomes less clear and emotions stronger.

9 – Extreme stress  
Close to panic or shutdown; functioning is very difficult.

10 – Maximum stress / crisis  
Panic, breakdown, or inability to cope.

How people usually measure it:
- Self‑check: “Where am I right now on the 1–10 scale?”
- Body cues: heart rate, muscle tension, breathing, headaches.
- Mental cues: racing thoughts, inability to focus, irritability.
- Behavior cues: avoidance, snapping at people, mistakes.

Many therapists and coaches ask clients to rate stress before and after an activity (meeting, workout, breathing exercise) to see what helps.

### Sweating Level (1–5)

1 – None / Dry  
Skin is dry. No noticeable sweating.

2 – Slight  
Very light moisture on skin (forehead, palms, or underarms), but not visible to others.

3 – Moderate  
Clearly noticeable sweat. Skin feels damp; small sweat spots may appear on clothing.

4 – Heavy  
Visible sweating and larger sweat patches on clothing. Sweat may drip or require wiping.

5 – Extreme / Profuse  
Constant dripping or soaking clothing. Sweat runs down face or body.

This type of scale is often used in exercise studies, heat stress monitoring, or medical observations to quickly rate sweating intensity.

---

### Diet Quality (1–10)

1 – Very poor  
Mostly ultra‑processed foods, sugary drinks, fast food, little to no fruits or vegetables.

2 – Poor  
Frequent junk food and refined carbs; very low nutrient variety.

3 – Below average  
Some real food, but still heavy in processed items and added sugars.

4 – Somewhat unhealthy  
Occasional fruits/vegetables but inconsistent balance; many refined foods.

5 – Average  
Mixed diet: some healthy meals, some unhealthy ones.

6 – Fairly good  
Regular fruits/vegetables and home‑cooked meals, but still room for improvement.

7 – Good  
Balanced meals most of the time; decent protein, whole grains, fruits, and vegetables.

8 – Very good  
Mostly whole foods, good nutrient balance, limited processed foods.

9 – Excellent  
Highly balanced diet with plenty of whole foods, healthy fats, lean proteins, fiber.

10 – Optimal  
Consistently nutrient‑dense, varied diet aligned with strong nutrition guidelines (lots of whole foods, minimal processed items).

People often estimate this based on factors like:
- fruit and vegetable intake  
- whole vs processed foods  
- sugar and sugary drink consumption  
- protein quality  
- variety of nutrients

### Difference Between Stress Levels and Anxiety Levels

Stress and anxiety are related but not the same.

Stress
- Usually a response to a specific external pressure (deadline, exam, argument, workload).
- Often temporary and improves when the situation passes.
- More tied to current demands.
- Symptoms: tension, irritability, trouble concentrating, headaches, fatigue.
- Example: feeling stressed because you have a project due tomorrow.

Anxiety
- More about worry, fear, or anticipation of possible threats, sometimes without a clear trigger.
- Can persist even when no immediate stressor exists.
- Often involves excessive worrying about the future.
- Symptoms: constant worry, restlessness, racing thoughts, rapid heartbeat, sleep problems.
- Example: worrying for days that you might fail something even when nothing immediate is happening.

Simple way to think about it
- Stress: pressure from a situation.  
- Anxiety: the internal feeling of fear or worry, sometimes continuing after the stressor is gone.

They often overlap: prolonged stress can lead to anxiety, and anxiety can make normal stress feel much stronger.

### Anxiety Level (1–10)

1 – No anxiety  
Calm, relaxed, no worry.

2 – Very mild  
Slight uneasiness but barely noticeable.

3 – Mild  
Occasional worry or nervousness, but it doesn’t affect functioning.

4 – Noticeable  
Anxiety is present and distracting at times.

5 – Moderate  
Frequent worry; concentration takes effort.

6 – Elevated  
Persistent anxious thoughts, some physical symptoms (restlessness, muscle tension).

7 – High  
Difficult to focus; clear physical symptoms like faster heartbeat or shallow breathing.

8 – Very high  
Strong worry, feeling overwhelmed, hard to think clearly.

9 – Severe  
Near panic; racing thoughts, strong physical discomfort.

10 – Extreme / panic level  
Panic attack or feeling completely unable to cope.

People often judge their number using:
- Thoughts: racing worries, fear, rumination  
- Body signals: heart rate, sweating, tension, nausea  
- Functioning: how much anxiety interferes with tasks or decisions.

### invalid values / out-of-range values

```python
# replace invalid values with NaN
def replace_invalid_values(data, col, condition):
    data = data.copy()
    n_invalid = condition.sum()
    data.loc[condition, col] = np.nan
    print(f'{col}: {n_invalid} invalid values converted to NaN')
    return data
```

```python
data = replace_invalid_values(data, 'Sleep Hours', data['Sleep Hours'] < 0)
data = replace_invalid_values(data, 'Physical Activity (hrs/week)', data['Physical Activity (hrs/week)'] < 0)
data = replace_invalid_values(data, 'Alcohol Consumption (drinks/week)', data['Alcohol Consumption (drinks/week)'] < 0)
data = replace_invalid_values(data, 'Stress Level (1-10)', data['Stress Level (1-10)'] > 10)
```

```python
# IQR-based lower/upper bounds for a numeric column
def outlier_bounds(data, col, k=1.5):
    q1, q3 = data[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr

low, high = outlier_bounds(data, 'Caffeine Intake (mg/day)')
n_outliers = ((data['Caffeine Intake (mg/day)'] < low) | (data['Caffeine Intake (mg/day)'] > high)).sum()
print(f'IQR bounds: [{low:.1f}, {high:.1f}]  ->  {n_outliers} potential outliers')
```

### Impute NaNs

```python
def fill_numeric(data, cols):
    data = data.copy()
    for c in cols:
        if data[c].isnull().sum() > 0:
            med = data[c].median()
            print(c, '->', data[c].isnull().sum(), 'filled with median', med)
            data[c] = data[c].fillna(med)
    return data

def fill_categorical(data, cols):
    data = data.copy()
    for c in cols:
        if data[c].isnull().sum() > 0:
            m = data[c].mode()[0]
            print(c, '->', data[c].isnull().sum(), 'filled with mode', m)
            data[c] = data[c].fillna(m)
    return data
```

```python
numeric_to_fill = ['Age', 'Sleep Hours', 'Physical Activity (hrs/week)',
                   'Caffeine Intake (mg/day)', 'Alcohol Consumption (drinks/week)',
                   'Stress Level (1-10)', 'Diet Quality (1-10)']

categorical_to_fill = ['Gender', 'Smoking', 'Medication']

data = fill_numeric(data, numeric_to_fill)
data = fill_categorical(data, categorical_to_fill)
```

## Outliers, Binning & Reclassifying

### Outliers

```python
def detect_outliers(col, k=1.5):
    q1, q3 = col.quantile(0.25), col.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr

    return (col < lower) | (col > upper)

def clip_outliers(col, k=1.5):
    q1, q3 = col.quantile(0.25), col.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr

    return col.clip(lower, upper)
```

```python
for c in ['Caffeine Intake (mg/day)', 'Heart Rate (bpm)', 'Breathing Rate (breaths/min)']:
    print(c, detect_outliers(data[c]).sum())
```

### Binning

Binning turns a continuous variable into a smaller number of meaningful groups — useful when the raw
scale is noisy but a coarser grouping tells a clearer story (e.g. age brackets instead of exact age).

```python
def binning(data, col, method='equal width', k=5, thresholds=None, labels=None):
    data = data.copy()
    new_col = col + '_bin'
    if method == 'equal width':
        data[new_col] = pd.cut(data[col], bins=k)
    elif method == 'custom':
        data[new_col] = pd.cut(data[col], bins=thresholds, labels=labels)
    else:
        raise ValueError("method must be 'equal width' or 'custom'")
    return data
```

```python
def binning(data, col, method='custom', bins=5, thresholds=None, labels=None):
    
    df = data.copy()
    
    new_col = col + "_bin"
    
    if method == "equal width":
        df[new_col] = pd.cut(df[col], bins=bins,  labels=labels)
    
    elif method == "quantile":
        df[new_col] = pd.qcut(df[col], q=bins,  labels=labels)
    
    elif method == "custom":
        df[new_col] = pd.cut(df[col], bins=thresholds, labels=labels)
    
    print("column:", new_col)
    print("number of groups:", df[new_col].nunique())
    
    return df
```

```python
data = binning(data, 'Age', method='custom', bins=4,
               thresholds=[18, 30, 45, 60, 65],
               labels=['18-29', '30-44', '45-59', '60-64'])

data['Age_bin'].value_counts().sort_index()
```

### Reclassifying categorical variables


```python
def reclassify_categorical(data, col, mapping, new_col=None):
    data = data.copy()
    new_col = new_col or f'{col}_recls'
    data[new_col] = data[col].map(mapping)
    return data
```

```python
occupation_groups = {
    'Doctor': 'Healthcare', 'Nurse': 'Healthcare',
    'Teacher': 'Education', 'Student': 'Education',
    'Artist': 'Creative', 'Musician': 'Creative',
    'Engineer': 'Technical', 'Scientist': 'Technical',
    'Lawyer': 'Professional', 'Freelancer': 'Professional',
    'Chef': 'Service', 'Athlete': 'Service',
    'Other': 'Other'
}

data = reclassify_categorical(data, 'Occupation', occupation_groups)
data['Occupation_recls'].value_counts()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=data, y='Occupation_recls',
              order=data['Occupation_recls'].value_counts().index, color='#55A868')
plt.title('Respondents per reclassified Occupation group')
plt.tight_layout()
plt.show()
```
