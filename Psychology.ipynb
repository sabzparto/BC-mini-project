### numeric columns
histogram+KDE to see the shape of the distribution, boxplot to spot outliers fast

```python
def plot_numeric(df, col):
    fig, axes = plt.subplots(1, 2, figsize=(10, 3))
    sns.histplot(df[col], kde=True, ax=axes[0])
    axes[0].set_title(col)
    sns.boxplot(x=df[col], ax=axes[1])
    axes[1].set_title('boxplot')
    plt.tight_layout()
    plt.show()
```

```python
for col in numeric_columns:
    if col in ['Target', 'is_Anxious']:
        continue
    plot_numeric(data, col)
```

### Stress Level (1–10)
A simple self‑rating tool where a person estimates how stressed they feel right now. It’s subjective, but using clear anchors helps keep ratings consistent.

1 – Completely relaxed  
No noticeable stress. Calm, clear thinking, body relaxed.

2 – Very mild tension  
Slight pressure but easily manageable.

3 – Low stress  
Aware of some demands, but still comfortable and functioning well.

4 – Noticeable stress  
You feel somewhat pressured or distracted.

5 – Moderate stress  
Stress is clearly present; concentration takes more effort.

6 – Elevated stress  
Feeling tense or irritable; performance may start to drop.

7 – High stress  
Difficult to focus, physical symptoms may appear (tight shoulders, faster heart rate).

8 – Very high stress  
Overwhelmed; thinking becomes less clear and emotions stronger.

9 – Extreme stress  
Close to panic or shutdown; functioning is very difficult.

10 – Maximum stress / crisis  
Panic, breakdown, or inability to cope.

How people usually measure it:
- Self‑check: “Where am I right now on the 1–10 scale?”
- Body cues: heart rate, muscle tension, breathing, headaches.
- Mental cues: racing thoughts, inability to focus, irritability.
- Behavior cues: avoidance, snapping at people, mistakes.

Many therapists and coaches ask clients to rate stress before and after an activity (meeting, workout, breathing exercise) to see what helps.

### Sweating Level (1–5)

1 – None / Dry  
Skin is dry. No noticeable sweating.

2 – Slight  
Very light moisture on skin (forehead, palms, or underarms), but not visible to others.

3 – Moderate  
Clearly noticeable sweat. Skin feels damp; small sweat spots may appear on clothing.

4 – Heavy  
Visible sweating and larger sweat patches on clothing. Sweat may drip or require wiping.

5 – Extreme / Profuse  
Constant dripping or soaking clothing. Sweat runs down face or body.

This type of scale is often used in exercise studies, heat stress monitoring, or medical observations to quickly rate sweating intensity.

### Difference Between Stress Levels and Anxiety Levels

Stress and anxiety are related but not the same.

Stress
- Usually a response to a specific external pressure (deadline, exam, argument, workload).
- Often temporary and improves when the situation passes.
- More tied to current demands.
- Symptoms: tension, irritability, trouble concentrating, headaches, fatigue.
- Example: feeling stressed because you have a project due tomorrow.

Anxiety
- More about worry, fear, or anticipation of possible threats, sometimes without a clear trigger.
- Can persist even when no immediate stressor exists.
- Often involves excessive worrying about the future.
- Symptoms: constant worry, restlessness, racing thoughts, rapid heartbeat, sleep problems.
- Example: worrying for days that you might fail something even when nothing immediate is happening.

Simple way to think about it
- Stress: pressure from a situation.  
- Anxiety: the internal feeling of fear or worry, sometimes continuing after the stressor is gone.

They often overlap: prolonged stress can lead to anxiety, and anxiety can make normal stress feel much stronger.

## Illogical entries

**sleep hours** contain illogical entries. \
the same applies to **Physical acitivity**. \
the same applies to **Alcohol Consumption**. \
**stress level** has max more than 10. \
any unmentioned column is fine.

before checking duplicates/nan/normality below, need to actually turn the illogical values into NaN first (otherwise normality test runs on garbage numbers like stress=15).

wrote these as generic functions instead of one per column, just pass in which columns have the problem

```python
def negative_to_nan(df, cols):
    # negative values dont make sense for these (cant sleep -10 hours)
    df = df.copy()
    for c in cols:
        bad = df[c] < 0
        print(c, '->', bad.sum(), 'negative values found')
        df.loc[bad, c] = np.nan
    return df
```

```python
def out_of_range_to_nan(df, ranges):
    # ranges = {col: (min, max)}
    df = df.copy()
    for c, (lo, hi) in ranges.items():
        bad = (df[c] < lo) | (df[c] > hi)
        print(c, '->', bad.sum(), 'values out of range')
        df.loc[bad, c] = np.nan
    return df
```

```python
def sentinel_to_nan(df, sentinels):
    # for weird constant values that show up way too often to be real
    # e.g. caffeine intake = exactly 1500 thirty times, heart rate = exactly 220 thirty times
    df = df.copy()
    for c, val in sentinels.items():
        bad = df[c] == val
        print(c, '->', bad.sum(), f'entries equal to {val}')
        df.loc[bad, c] = np.nan
    return df
```

```python
data = negative_to_nan(data, ['Sleep Hours', 'Physical Activity (hrs/week)', 'Alcohol Consumption (drinks/week)'])
data = out_of_range_to_nan(data, {'Stress Level (1-10)': (1, 10)})
data = sentinel_to_nan(data, {'Caffeine Intake (mg/day)': 1500.0, 'Heart Rate (bpm)': 220})
```

now data has proper NaNs instead of fake numbers, rest of the checks below use this

```python
def nan_percentage(df : DataFrame) -> DataFrame:
    total_rows = len(df)
    nan_percentage = [(len(df[df[column].isna()]) / total_rows) * 100 for column in df.columns]
    result = DataFrame(nan_percentage, index = df.columns  , columns = ['Nan Percent'])
    return result
```

## Normality

```python
# this function checks the normality by deleting the nan
def normality_checker(df : DataFrame , alpha : float = 0.05) -> None:
    normal_col = []
    for column in df.select_dtypes(include=['number']).columns:
        if shapiro(df[column] , nan_policy = 'omit')[1] > alpha:
            normal_col.append(column)
    if normal_col:
        print('List of columns with normal distribution:' , normal_col , sep = '\n')
    else:
        print('No columns follow a normal distribution.')
```

```python
normality_checker(data)
```

#### missing values

for numbers: if the column is roughly normal use mean, otherwise median (basically always median here, normality_checker already said nothing is normal)

for categories: fill with the most common value, except Therapy History which is like 90% missing so filling with mode would basically be making stuff up -> gets its own 'Unknown' category instead

doing this before the outlier step below since that one needs no NaNs left (more on that there)

```python
def col_is_normal(s, alpha=0.05):
    s = s.dropna()
    if len(s) > 5000:
        s = s.sample(5000, random_state=1)
    return shapiro(s)[1] > alpha


def fill_numeric_na(df, cols):
    df = df.copy()
    for c in cols:
        if df[c].isna().sum() == 0:
            continue
        if col_is_normal(df[c]):
            val = df[c].mean()
            print(c, '- filled with mean:', round(val,2))
        else:
            val = df[c].median()
            print(c, '- filled with median:', round(val,2))
        df[c] = df[c].fillna(val)
    return df


def fill_categorical_na(df, cols, missing_cutoff=0.5):
    df = df.copy()
    for c in cols:
        n_missing = df[c].isna().sum()
        if n_missing == 0:
            continue
        if n_missing / len(df) > missing_cutoff:
            print(c, '- mostly missing, filling with Unknown')
            df[c] = df[c].fillna('Unknown')
        else:
            mode = df[c].mode()[0]
            print(c, '- filled with mode:', mode)
            df[c] = df[c].fillna(mode)
    return df
```
