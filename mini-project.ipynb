### Statistical Inference Functions

#### Frequency Table

```python
#frequency: create table to show 

def freq_table(column):
    counts = data[column].value_counts()
    percent = (counts / counts.sum()) * 10

    table = pd.DataFrame({'counts' : counts, 'percent' : percent})

    return table
```

#### Correlation

```python
#محاسبه همبستگی سایر متغیرها با مضطرب بودن
def correlation_with_target(data, columns, target):
    
    for col in columns:
        if col == target:
            continue
        
        corr, p = stats.spearmanr(data[col], data[target])
    
        
        print(col)
        print("corr :", corr)
        print("p :", p)
        
        
        if p < 0.05:
            print("significant: فرض صفر رد می‌شود")
        else:
            print("not significant: فرض صفر رد نمی‌شود")
       
```

```python
def correlation_matrix(data, columns):
    corr = data[columns].corr(method="spearman")
    return corr
```

```python
def one_sample_test(data, column, value):
    col_data = data[column]
    col_data = col_data.dropna()
    
    result = stats.ttest_1samp(col_data, value)
    stat = result.statistic
    p = result.pvalue
    
    print("mean of sample :", col_data.mean())
    print("t :", stat)
    print("p :", p)
    
    if p < 0.05:
        print("فرض صفر رد می‌شود", value)
    else:
        print(" فرض صفر رد نمی‌شود", value)
```
